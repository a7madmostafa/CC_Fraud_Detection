# Import Libraries

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd 
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.compose import ColumnTransformer
from imblearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from imblearn.over_sampling import SMOTE

RANDOM_STATE = 42
SCORING_METRIC = "average_precision"
```

# Data Split

```python
X = df.drop('Class', axis=1)
y = df['Class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)
```

```python
y_train.value_counts().round(3)
```

```python
y_test.value_counts(normalize=True).round(3)
```

# Preprocessing

```python
scaler = StandardScaler()
log_transformer = FunctionTransformer(np.log1p, inverse_func = np.expm1)

other_cols = X.columns.drop('Amount')

amount_pipe = Pipeline([
    ("log", FunctionTransformer(np.log1p)),
    ("scale", StandardScaler()),
])

prep = ColumnTransformer([
    ("amount", amount_pipe, ['Amount']),
    ("rest", StandardScaler(), other_cols),
])
```

# Base Model

```python
log_reg = Pipeline([
                     ("prep", prep),
                     ("model", LogisticRegression())
])

kf = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)
param_grid = {
    "model__C": [0.01, 1, 10],
    "model__class_weight": [None, "balanced"]
}

log_search = GridSearchCV(log_reg, param_grid, cv=kf, scoring=SCORING_METRIC)
log_search.fit(X_train, y_train)
```

```python
log_search.best_params_
```

```python
log_search.best_score_
```

# Try SMOTE

```python
log_reg_smote = Pipeline([
                     ("prep", prep),
                     ("sampler", SMOTE(random_state=RANDOM_STATE)),
                     ("model", LogisticRegression(max_iter=1000))
])

param_grid_smote = {
    "sampler__sampling_strategy": [0.1, 0.2],
    "model__C": [0.01, 1, 10],
}

smote_search = GridSearchCV(log_reg_smote, param_grid_smote, cv=kf, scoring=SCORING_METRIC, n_jobs=-1)
smote_search.fit(X_train, y_train)

print(smote_search.best_params_)
print("SMOTE CV score   :", smote_search.best_score_)
print("No-SMOTE CV score:", log_search.best_score_)
```

# Decision Tree

```python
tree_smote = Pipeline([
                     ("prep", prep),
                     ("sampler", SMOTE(random_state=RANDOM_STATE)),
                     ("model", DecisionTreeClassifier(random_state=RANDOM_STATE))
])

tree_param_grid = {
    "sampler__sampling_strategy": [0.1, 0.2],
    "model__max_depth": [5, 10, 20],
    'model__min_samples_leaf': [100, 150],
}

tree_search = GridSearchCV(tree_smote, tree_param_grid, cv=kf, scoring=SCORING_METRIC, n_jobs=-1)
tree_search.fit(X_train, y_train)

print(tree_search.best_params_)
print("SMOTE CV score   :", tree_search.best_score_)
```
